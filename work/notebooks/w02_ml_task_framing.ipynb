## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

For the starter dataset, I will use a **derived decline proxy** rather than claiming to have a future observed target. I define `is_declining_proxy` as `trend_direction == "down"`. This label represents the dataset's existing definition of a declining item and is derived from `trend_pct`, so it is not an independently observed future outcome.

The proxy allows me to investigate whether other available content and performance signals are associated with the existing decline definition. `trend_pct` and `trend_direction` will not be used as predictive features because they are part of the proxy's derivation. In later work, I will investigate constructing a stronger target from the warehouse using a future performance window.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Load starter dataset
df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

# Create the provisional decline proxy
df["is_declining_proxy"] = df["trend_direction"] == "down"

print("Proxy definition:")
print("is_declining_proxy = (trend_direction == 'down')")

print("\nProxy distribution:")
print(df["is_declining_proxy"].value_counts())

print("\nProxy rate:")
print(f"{df['is_declining_proxy'].mean():.2%}")
```
