## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter dataset contains 30,000 content items across 32 pseudonymized clients, providing enough scale and cross-client coverage to investigate a content-prioritization problem.

I define a provisional binary decline proxy from the dataset's `trend_direction`: an item is marked as declining when `trend_direction == "down"`. This produces [X] declining items ([Y]%) in the starter dataset.

The dataset also contains 1,205 items where `avg_position = 0`. According to the data dictionary, this represents unavailable GSC position data rather than a real search position. This highlights an important missing-data issue that the eventual modeling pipeline will need to handle explicitly.

```python
import pandas as pd

# Load starter dataset
df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

# Dataset scale
total_items = len(df)
num_clients = df["client_id"].nunique()

# Provisional decline proxy
df["is_declining_label"] = df["trend_direction"] == "down"

declining_count = df["is_declining_label"].sum()
declining_pct = declining_count / total_items * 100

# GSC availability
missing_gsc = (df["avg_position"] == 0).sum()
missing_gsc_pct = missing_gsc / total_items * 100

print("=== STARTER DATASET PROOF POINTS ===")
print(
    f"1. Dataset scale: {total_items:,} content items "
    f"across {num_clients} pseudonymized clients."
)
print(
    f"2. Decline proxy: {declining_count:,} items "
    f"({declining_pct:.2f}%) have trend_direction='down'."
)
print(
    f"3. GSC availability: {missing_gsc:,} items "
    f"({missing_gsc_pct:.2f}%) have avg_position=0."
)
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

### What I Can Claim
* I can measure associations between available content/performance signals and the starter dataset's derived decline proxy (trend_direction == "down").
* I can evaluate whether a ranking approach can provide decision-support by prioritizing content items for editorial review, using metrics such as Precision@K.
* I can compare the ranking/model against simple heuristic baselines and report whether the model provides a measured improvement rather than assuming ML is better.
* With the warehouse data, I can investigate whether historical signals have directional value for a later observed performance outcome, if a valid future-window target can be constructed.
* I can identify signals that are associated with declining performance under the defined evaluation setup, without interpreting those associations as causal effects.

### What I Cannot Claim
* **No causal claims:** I cannot claim that changing a page, feature, or content characteristic will cause its search or traffic performance to improve.
* **No "predicting Google":** I cannot claim that the model predicts Google's ranking algorithm, future algorithm updates, or search rankings with certainty.
* **No guaranteed traffic recovery:** A page receiving a high priority score does not mean that refreshing it will recover its traffic or search position.
* **No automatic ML superiority:** I cannot claim that ML is better than a fixed rule unless comparison against appropriate baselines demonstrates this empirically.
* **No future-prediction claim from the starter proxy:** is_declining_label, which I derive from trend_direction == "down", represents the dataset's existing definition of decline. Results using it will be described as associations with the decline proxy, not as evidence that the model predicts future traffic decline.
* **No target leakage:** trend_pct and trend_direction will not be used as predictive features when the derived is_declining_label is the target, because both are part of the target's derivation.
