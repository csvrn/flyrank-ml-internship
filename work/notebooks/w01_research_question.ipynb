```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
import numpy as np

# Load starter dataset
df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')

# Metric 1: Dataset Scale & Baseline Base Rate
total_items = len(df)
num_clients = df['client_id'].nunique()
df['is_declining_label'] = df['trend_direction'] == 'down'
declining_count = df['is_declining_label'].sum()
declining_pct = (declining_count / total_items) * 100

# Metric 2: GSC Missingness Signal (avg_position = 0 trap)
missing_gsc = (df['avg_position'] == 0).sum()
missing_gsc_pct = (missing_gsc / total_items) * 100

# Metric 3: Domain Heterogeneity across Clients
top_client_share = (df['client_id'].value_counts().iloc[0] / total_items) * 100

print(f"=== STARTER DATASET PROOF POINTS ===")
print(f"1. Problem Scale: {total_items:,} content items across {num_clients} client domains.")
print(f"   - Baseline decay rate: {declining_count:,} items ({declining_pct:.2f}%) marked as declining.")
print(f"2. Data Integrity Signal: {missing_gsc:,} items ({missing_gsc_pct:.2f}%) have avg_position = 0 (missing GSC data).")
print(f"3. Client Distribution: Largest client holds {top_client_share:.2f}% of total rows, confirming multi-tenant variance.")
```
